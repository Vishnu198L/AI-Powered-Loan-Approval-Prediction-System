```python
# Check missing values in every column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values[missing_values > 0])

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
print("Applicants older than 100:")
display(df[df["person_age"] > 100])

print("Employment length above 50 years:")
display(df[df["person_emp_length"] > 50])
```

```python
# Remove unrealistic age and employment length values

df = df[
    (df["person_age"] <= 100) &
    (
        (df["person_emp_length"] <= 50) |
        (df["person_emp_length"].isna())
    )
].copy()

print("Dataset shape after cleaning:", df.shape)
```

```python
from pathlib import Path

# Create the processed-data folder
processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

# Save the cleaned dataset
output_path = processed_dir / "loan_approval_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", output_path.resolve())
```
